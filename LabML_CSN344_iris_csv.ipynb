To display images in Python, you can use libraries like `Pillow` (PIL) for image manipulation and `matplotlib.pyplot` for plotting. First, let's download an example image from the internet.

```python
import requests
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt

# URL of an example image (updated to a different, publicly accessible image)
image_url = 'https://www.google.com/images/branding/googlelogo/2x/googlelogo_color_272x92dp.png'

# Download the image
response = requests.get(image_url)
response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
img_data = response.content

# Print content type and first few bytes for debugging
print(f"Content-Type: {response.headers.get('Content-Type')}")
print(f"First 50 bytes of data: {img_data[:50]}")

# Open the image using Pillow
img = Image.open(BytesIO(img_data))

print(f"Image size: {img.width}x{img.height}")
```

Now that the image is loaded, you can display it using `matplotlib.pyplot`.

```python
# Display the image
plt.imshow(img)
plt.axis('off') # Hide axes
plt.title('Example Image')
plt.show()
```

### Data Visualizations from Iris Dataset

Below are 5 different plots generated from the `iris` dataset, serving as 'sample images' to visualize its characteristics.

```python
# 1. Scatter plot of Sepal Length vs. Sepal Width, colored by species
plt.figure(figsize=(8, 6))
sns.scatterplot(x='sepal_length', y='sepal_width', hue='species', data=df, s=100)
plt.title('Sepal Length vs. Sepal Width by Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# 2. Scatter plot of Petal Length vs. Petal Width, colored by species
plt.figure(figsize=(8, 6))
sns.scatterplot(x='petal_length', y='petal_width', hue='species', data=df, s=100)
plt.title('Petal Length vs. Petal Width by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# 3. Histogram of Sepal Length for each species
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='sepal_length', hue='species', multiple='stack', kde=True)
plt.title('Distribution of Sepal Length by Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# 4. Histogram of Petal Length for each species
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='petal_length', hue='species', multiple='stack', kde=True)
plt.title('Distribution of Petal Length by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# 5. Box plot of all features by species
plt.figure(figsize=(12, 8))
sns.boxplot(data=df.drop('species', axis=1))
plt.title('Box Plot of Iris Features')
plt.ylabel('Measurement (cm)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Alternatively, box plots for each feature grouped by species
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
for i, feature in enumerate(features):
    row = i // 2
    col = i % 2
    sns.boxplot(x='species', y=feature, data=df, ax=axes[row, col])
    axes[row, col].set_title(f'Distribution of {feature} by Species')
    axes[row, col].grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Data Visualization: Heatmaps

Heatmaps are a great way to visualize the correlation matrix between numerical features and to see feature distributions across different categories.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Correlation Heatmap
# Calculate the correlation matrix for numerical features
correlation_matrix = df.drop('species', axis=1).corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Heatmap of Iris Features')
plt.show()
```

```python
# 2. Heatmap of mean feature values by Species
# Calculate the mean of each numerical feature grouped by species
mean_features_by_species = df.groupby('species').mean()

plt.figure(figsize=(10, 5))
sns.heatmap(mean_features_by_species, annot=True, cmap='viridis', fmt=".2f", linewidths=.5)
plt.title('Mean Feature Values by Species')
plt.xlabel('Features')
plt.ylabel('Species')
plt.show()
```
