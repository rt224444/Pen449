```python

# ============================================================
# ICE CREAM SENSORY EVALUATION + MACHINE LEARNING EXPERIMENT
# Synthetic 1,000-observation dataset
# ============================================================

# 1. Install/import libraries
!pip -q install openpyxl seaborn

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier, GradientBoostingRegressor
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report, roc_auc_score
)
from sklearn.inspection import permutation_importance

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

# ------------------------------------------------------------
# 2. Upload dataset to Google Colab
# ------------------------------------------------------------
from google.colab import files

uploaded = files.upload()

# If the uploaded filename differs, change this line.
DATA_FILE = "/content/ice_cream_sensory_dataset_1000.csv"
df = pd.read_csv(DATA_FILE)

print("Dataset shape:", df.shape)
display(df.head())
display(df.info())

# ------------------------------------------------------------
# 3. Basic data-quality checks
# ------------------------------------------------------------
print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False).head(10))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nUnique participants:", df["Participant_ID"].nunique())
print("Unique formulations:", df["Sample_ID"].nunique())

print("\nDataset split:")
display(df["Dataset_Split"].value_counts())

# ------------------------------------------------------------
# 4. Descriptive statistics
# ------------------------------------------------------------
sensory_cols = [
    "Appearance", "Color", "Aroma", "Sweetness", "Creaminess",
    "Flavor_Intensity", "Texture", "Mouthfeel", "Aftertaste",
    "Temperature_Acceptance", "Overall_Acceptance", "Satisfaction"
]

display(df[sensory_cols].describe().T)

# ------------------------------------------------------------
# 5. Distribution of overall acceptance
# ------------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.histplot(df["Overall_Acceptance"], bins=9, kde=True)
plt.title("Distribution of Overall Ice-Cream Acceptance")
plt.xlabel("Overall Acceptance (1–9)")
plt.ylabel("Number of Tasting Observations")
plt.show()

# ------------------------------------------------------------
# 6. Overall acceptance by flavor
# ------------------------------------------------------------
flavor_summary = (
    df.groupby("Flavor")["Overall_Acceptance"]
      .agg(["count", "mean", "std"])
      .sort_values("mean", ascending=False)
)

display(flavor_summary)

plt.figure(figsize=(11, 5))
sns.boxplot(data=df, x="Flavor", y="Overall_Acceptance")
plt.xticks(rotation=45, ha="right")
plt.title("Overall Acceptance by Ice-Cream Flavor")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. Sensory correlation matrix
# ------------------------------------------------------------
corr_cols = [
    "Appearance", "Color", "Aroma", "Sweetness", "Creaminess",
    "Flavor_Intensity", "Texture", "Mouthfeel", "Aftertaste",
    "Temperature_Acceptance", "Overall_Acceptance", "Satisfaction"
]

plt.figure(figsize=(12, 9))
sns.heatmap(
    df[corr_cols].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)
plt.title("Correlation Matrix of Sensory Attributes")
plt.tight_layout()
plt.show()

# ============================================================
# 8. MACHINE-LEARNING TARGETS
# ============================================================
# Primary regression target:
#   Overall_Acceptance
#
# Classification target:
#   Preference_Class
#
# IMPORTANT:
# Satisfaction, Purchase_Intent and Preference_Class are outcomes
# and are therefore excluded from the predictor set.

feature_cols = [
    "Age_Group",
    "Gender",
    "Previous_Ice_Cream_Consumption",
    "Tasting_Environment",
    "Flavor",
    "Base_Type",
    "Fat_Level_pct",
    "Sugar_Level_pct",
    "Fruit_Content_pct",
    "Stabilizer_Level_pct",
    "Appearance",
    "Color",
    "Aroma",
    "Sweetness",
    "Creaminess",
    "Flavor_Intensity",
    "Texture",
    "Mouthfeel",
    "Aftertaste",
    "Temperature_Acceptance"
]

X = df[feature_cols]
y_reg = df["Overall_Acceptance"]
y_clf = df["Preference_Class"]

# ------------------------------------------------------------
# 9. Use the pre-created participant-level split
# ------------------------------------------------------------
train_mask = df["Dataset_Split"].eq("Train")
val_mask = df["Dataset_Split"].eq("Validation")
test_mask = df["Dataset_Split"].eq("Test")

X_train, y_reg_train = X.loc[train_mask], y_reg.loc[train_mask]
X_val, y_reg_val = X.loc[val_mask], y_reg.loc[val_mask]
X_test, y_reg_test = X.loc[test_mask], y_reg.loc[test_mask]

y_clf_train = y_clf.loc[train_mask]
y_clf_val = y_clf.loc[val_mask]
y_clf_test = y_clf.loc[test_mask]

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)

# ------------------------------------------------------------
# 10. Preprocessing pipeline
# ------------------------------------------------------------
categorical_features = [
    "Age_Group", "Gender", "Previous_Ice_Cream_Consumption",
    "Tasting_Environment", "Flavor", "Base_Type"
]

numeric_features = [
    "Fat_Level_pct", "Sugar_Level_pct", "Fruit_Content_pct",
    "Stabilizer_Level_pct", "Appearance", "Color", "Aroma",
    "Sweetness", "Creaminess", "Flavor_Intensity", "Texture",
    "Mouthfeel", "Aftertaste", "Temperature_Acceptance"
]

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# ============================================================
# 11. REGRESSION MODEL 1 — Random Forest
# ============================================================
rf_reg = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=400,
        max_depth=None,
        min_samples_leaf=2,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ))
])

rf_reg.fit(X_train, y_reg_train)

val_pred_rf = rf_reg.predict(X_val)
test_pred_rf = rf_reg.predict(X_test)

def regression_report(name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f"\n{name}")
    print(f"MAE  : {mae:.3f}")
    print(f"RMSE : {rmse:.3f}")
    print(f"R²   : {r2:.3f}")
    return [name, mae, rmse, r2]

reg_results = []
reg_results.append(
    regression_report("Random Forest - Validation", y_reg_val, val_pred_rf)
)

reg_results.append(
    regression_report("Random Forest - Test", y_reg_test, test_pred_rf)
)

# ============================================================
# 12. REGRESSION MODEL 2 — Gradient Boosting
# ============================================================
gb_reg = Pipeline([
    ("preprocessor", preprocessor),
    ("model", GradientBoostingRegressor(
        n_estimators=250,
        learning_rate=0.04,
        max_depth=3,
        random_state=RANDOM_STATE
    ))
])

gb_reg.fit(X_train, y_reg_train)

val_pred_gb = gb_reg.predict(X_val)
test_pred_gb = gb_reg.predict(X_test)

reg_results.append(
    regression_report("Gradient Boosting - Validation", y_reg_val, val_pred_gb)
)

reg_results.append(
    regression_report("Gradient Boosting - Test", y_reg_test, test_pred_gb)
)

reg_results_df = pd.DataFrame(
    reg_results,
    columns=["Model", "MAE", "RMSE", "R2"]
)

display(reg_results_df)

# ------------------------------------------------------------
# 13. Actual vs predicted plot
# ------------------------------------------------------------
plt.figure(figsize=(7, 6))
plt.scatter(y_reg_test, test_pred_rf, alpha=0.65)
plt.plot([1, 9], [1, 9], linestyle="--")
plt.xlabel("Actual Overall Acceptance")
plt.ylabel("Predicted Overall Acceptance")
plt.title("Random Forest: Actual vs Predicted Acceptance")
plt.show()

# ============================================================
# 14. CLASSIFICATION — Preference Class
# ============================================================
rf_clf = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=400,
        min_samples_leaf=2,
        random_state=RANDOM_STATE,
        class_weight="balanced",
        n_jobs=-1
    ))
])

rf_clf.fit(X_train, y_clf_train)

val_pred_class = rf_clf.predict(X_val)
test_pred_class = rf_clf.predict(X_test)

print("\nRandom Forest Classification - Validation")
print(classification_report(y_clf_val, val_pred_class))

print("\nRandom Forest Classification - Test")
print(classification_report(y_clf_test, test_pred_class))

accuracy = accuracy_score(y_clf_test, test_pred_class)
precision = precision_score(
    y_clf_test, test_pred_class,
    average="weighted", zero_division=0
)
recall = recall_score(
    y_clf_test, test_pred_class,
    average="weighted", zero_division=0
)
f1 = f1_score(
    y_clf_test, test_pred_class,
    average="weighted", zero_division=0
)

classification_results = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "F1-Score"],
    "Test_Value": [accuracy, precision, recall, f1]
})

display(classification_results)

# ------------------------------------------------------------
# 15. Confusion matrix
# ------------------------------------------------------------
labels = ["Low", "Moderate", "High"]

cm = confusion_matrix(y_clf_test, test_pred_class, labels=labels)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=labels,
    yticklabels=labels,
    cmap="Blues"
)
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("Preference-Class Confusion Matrix")
plt.show()

# ============================================================
# 16. Logistic Regression baseline
# ============================================================
logistic_clf = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        max_iter=2000,
        multi_class="auto",
        class_weight="balanced",
        random_state=RANDOM_STATE
    ))
])

logistic_clf.fit(X_train, y_clf_train)

logistic_pred = logistic_clf.predict(X_test)

print("Logistic Regression Test Results")
print(classification_report(y_clf_test, logistic_pred))

logistic_accuracy = accuracy_score(y_clf_test, logistic_pred)
logistic_f1 = f1_score(
    y_clf_test, logistic_pred,
    average="weighted",
    zero_division=0
)

baseline_results = pd.DataFrame({
    "Model": ["Random Forest", "Logistic Regression"],
    "Accuracy": [accuracy, logistic_accuracy],
    "Weighted_F1": [f1, logistic_f1]
})

display(baseline_results)

# ============================================================
# 17. Feature importance using permutation importance
# ============================================================
# This operates on the original feature columns, so the result
# is easier to interpret than one-hot encoded feature names.

perm = permutation_importance(
    rf_reg,
    X_test,
    y_reg_test,
    n_repeats=10,
    random_state=RANDOM_STATE,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance": perm.importances_mean
}).sort_values("Importance", ascending=False)

display(importance_df)

plt.figure(figsize=(9, 7))
top = importance_df.head(12).sort_values("Importance")
plt.barh(top["Feature"], top["Importance"])
plt.xlabel("Permutation Importance")
plt.title("Top Predictors of Overall Acceptance")
plt.tight_layout()
plt.show()

# ============================================================
# 18. Purchase-intention analysis
# ============================================================
purchase_summary = (
    df.groupby("Preference_Class")["Purchase_Intent"]
      .agg(["count", "mean"])
      .rename(columns={"mean": "Purchase_Intent_Rate"})
)

display(purchase_summary)

plt.figure(figsize=(7, 5))
sns.barplot(
    data=df,
    x="Preference_Class",
    y="Purchase_Intent"
)
plt.ylabel("Purchase Intent Rate")
plt.title("Purchase Intent by Preference Class")
plt.show()

# ============================================================
# 19. New sample prediction
# ============================================================
new_sample = pd.DataFrame([{
    "Age_Group": "21-30",
    "Gender": "Male",
    "Previous_Ice_Cream_Consumption": "High",
    "Tasting_Environment": "Controlled sensory room",
    "Flavor": "Chocolate",
    "Base_Type": "Dairy",
    "Fat_Level_pct": 12.0,
    "Sugar_Level_pct": 16.0,
    "Fruit_Content_pct": 0.0,
    "Stabilizer_Level_pct": 0.35,
    "Appearance": 7.5,
    "Color": 7.5,
    "Aroma": 7.0,
    "Sweetness": 6.5,
    "Creaminess": 7.5,
    "Flavor_Intensity": 8.0,
    "Texture": 7.5,
    "Mouthfeel": 7.5,
    "Aftertaste": 7.0,
    "Temperature_Acceptance": 7.0
}])

predicted_acceptance = rf_reg.predict(new_sample)[0]
predicted_preference = rf_clf.predict(new_sample)[0]

print(f"Predicted Overall Acceptance: {predicted_acceptance:.2f}/9")
print(f"Predicted Preference Class: {predicted_preference}")

# ============================================================
# 20. Save model results
# ============================================================
results_path = "ice_cream_ml_results.csv"
reg_results_df.to_csv(results_path, index=False)

classification_results.to_csv(
    "ice_cream_classification_results.csv",
    index=False
)

importance_df.to_csv(
    "ice_cream_feature_importance.csv",
    index=False
)

print("Analysis files saved.")

# ============================================================
# 21. Research interpretation template
# ============================================================
print("""
RESEARCH INTERPRETATION
-----------------------
The analysis evaluates whether sensory attributes can predict
consumer acceptance and preference class.

Regression:
- MAE measures the average absolute prediction error.
- RMSE penalizes larger errors more strongly.
- R² describes the proportion of variation explained by the model.

Classification:
- Accuracy measures overall correct classification.
- Precision measures reliability of predicted classes.
- Recall measures how many observations in each class are recovered.
- F1-score balances precision and recall.

Methodological note:
The dataset is synthetic. It is appropriate for demonstrating
data science, machine-learning, pipeline development and
experimental methodology, but it must not be presented as
observations collected from real human participants.
""")

```

## 22. Neural Network Implementation (PyTorch Regression)

Let's implement a simple feedforward neural network using PyTorch to predict `Overall_Acceptance` and visualize its training loss over epochs.

```python
# 22.1 Import PyTorch libraries
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Ensure reproducibility
torch.manual_seed(RANDOM_STATE)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_STATE)

# Define device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
# 22.2 Prepare data for PyTorch

# Apply the preprocessor to the feature data
X_train_processed = preprocessor.transform(X_train)
X_val_processed = preprocessor.transform(X_val)
X_test_processed = preprocessor.transform(X_test)

# Convert to PyTorch tensors
X_train_tensor = torch.tensor(X_train_processed, dtype=torch.float32)
y_reg_train_tensor = torch.tensor(y_reg_train.values, dtype=torch.float32).view(-1, 1)

X_val_tensor = torch.tensor(X_val_processed, dtype=torch.float32)
y_reg_val_tensor = torch.tensor(y_reg_val.values, dtype=torch.float32).view(-1, 1)

X_test_tensor = torch.tensor(X_test_processed, dtype=torch.float32)
y_reg_test_tensor = torch.tensor(y_reg_test.values, dtype=torch.float32).view(-1, 1)

# Create TensorDatasets
train_dataset = TensorDataset(X_train_tensor, y_reg_train_tensor)
val_dataset = TensorDataset(X_val_tensor, y_reg_val_tensor)
test_dataset = TensorDataset(X_test_tensor, y_reg_test_tensor)

# Create DataLoaders
batch_size = 32
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

print(f"Number of features after preprocessing: {X_train_tensor.shape[1]}")
print(f"Train tensor shapes: {X_train_tensor.shape}, {y_reg_train_tensor.shape}")
```

```python
# 22.3 Define the Neural Network Architecture for Regression

class RegressionNN(nn.Module):
    def __init__(self, input_size):
        super(RegressionNN, self).__init__()
        self.fc1 = nn.Linear(input_size, 128)
        self.relu1 = nn.ReLU()
        self.dropout1 = nn.Dropout(0.3)
        self.fc2 = nn.Linear(128, 64)
        self.relu2 = nn.ReLU()
        self.dropout2 = nn.Dropout(0.3)
        self.fc3 = nn.Linear(64, 1) # Output a single regression value

    def forward(self, x):
        x = self.dropout1(self.relu1(self.fc1(x)))
        x = self.dropout2(self.relu2(self.fc2(x)))
        x = self.fc3(x)
        return x

# Instantiate the model
input_size = X_train_tensor.shape[1]
model = RegressionNN(input_size).to(device)
print(model)
```

```python
# 22.4 Define Loss Function, Optimizer, and Training Loop

criterion = nn.MSELoss() # Mean Squared Error for regression
optimizer = optim.Adam(model.parameters(), lr=0.001)

def train_model(model, train_loader, val_loader, criterion, optimizer, num_epochs=50):
    train_losses = []
    val_losses = []

    for epoch in range(num_epochs):
        model.train() # Set model to training mode
        running_train_loss = 0.0
        for inputs, targets in train_loader:
            inputs, targets = inputs.to(device), targets.to(device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()
            running_train_loss += loss.item() * inputs.size(0)

        epoch_train_loss = running_train_loss / len(train_loader.dataset)
        train_losses.append(epoch_train_loss)

        # Validation phase
        model.eval() # Set model to evaluation mode
        running_val_loss = 0.0
        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs, targets = inputs.to(device), targets.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, targets)
                running_val_loss += loss.item() * inputs.size(0)

        epoch_val_loss = running_val_loss / len(val_loader.dataset)
        val_losses.append(epoch_val_loss)

        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {epoch_train_loss:.4f}, Val Loss: {epoch_val_loss:.4f}')

    return train_losses, val_losses

# Train the model
num_epochs = 100
train_losses, val_losses = train_model(model, train_loader, val_loader, criterion, optimizer, num_epochs=num_epochs)
```

```python
# 22.5 Plot Loss vs. Epochs
plt.figure(figsize=(10, 6))
plt.plot(range(1, num_epochs + 1), train_losses, label='Training Loss')
plt.plot(range(1, num_epochs + 1), val_losses, label='Validation Loss')
plt.title('DNeural Network Loss vs. Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 22.6 Evaluate the model on the test set
model.eval()
all_predictions = []
all_targets = []

with torch.no_grad():
    for inputs, targets in test_loader:
        inputs, targets = inputs.to(device), targets.to(device)
        outputs = model(inputs)
        all_predictions.extend(outputs.cpu().numpy())
        all_targets.extend(targets.cpu().numpy())

all_predictions = np.array(all_predictions).flatten()
all_targets = np.array(all_targets).flatten()

# Calculate regression metrics
mae = mean_absolute_error(all_targets, all_predictions)
rmse = np.sqrt(mean_squared_error(all_targets, all_predictions))
r2 = r2_score(all_targets, all_predictions)

print(f"\nNeural Network Regression - Test Results:")
print(f"MAE  : {mae:.3f}")
print(f"RMSE : {rmse:.3f}")
print(f"R²   : {r2:.3f}")
```
